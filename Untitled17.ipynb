```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

url = 'https://raw.githubusercontent.com/allisonhorst/palmerpenguins/main/inst/extdata/penguins.csv'


print("Generating a highly realistic hourly traffic volume dataset with typical smart city patterns...")
np.random.seed(42)
timestamps = pd.date_range(start="2023-01-01", periods=3000, freq="h")


hours = timestamps.hour
traffic_pattern = 1000 + 1500 * np.sin(2 * np.pi * hours / 24) + 800 * np.sin(4 * np.pi * (hours - 8) / 24) + np.random.normal(0, 300, len(timestamps))
traffic_volume = np.clip(traffic_pattern, 100, 5000).astype(int)

df = pd.DataFrame({'date_time': timestamps, 'traffic_volume': traffic_volume})
print(f"Dataset successfully created. Shape: {df.shape}")

df.set_index('date_time', inplace=True)
subset_df = df.iloc[0:2500].copy()
print(f"Selected subset shape: {subset_df.shape}")


plt.figure(figsize=(15, 5))
plt.plot(subset_df.index, subset_df['traffic_volume'], label='Hourly Traffic Volume (Simulated)', color='blue')
plt.title('Hourly Traffic Volume for Smart City Forecasting (2,500 Consecutive Records)')
plt.xlabel('Date Time')
plt.ylabel('Traffic Volume (Vehicles/Hour)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
def create_sequential_windows(data, input_width, label_width, shift):
    """
    Generates sequential input and output windows while maintaining chronological order.

    Parameters:
    - data: 1D array-like or DataFrame column.
    - input_width: Number of historical lag steps (input window size).
    - label_width: Number of future steps to forecast (output window size).
    - shift: Offset representing steps between the start of input and start of label.
             Typically shift = input_width to forecast immediately after input.
    """

    values = np.array(data, dtype=np.float32)

    total_window_size = input_width + (shift - input_width) + label_width

    X = []
    y = []

    for i in range(len(values) - total_window_size + 1):
        input_window = values[i : i + input_width]
        label_window = values[i + input_width + (shift - input_width) : i + total_window_size]
        X.append(input_window)
        y.append(label_window)

    return np.array(X), np.array(y)


INPUT_WIDTH = 24
LABEL_WIDTH = 1
SHIFT = 24

X, y = create_sequential_windows(subset_df['traffic_volume'], INPUT_WIDTH, LABEL_WIDTH, SHIFT)

print(f"Original series length: {len(subset_df)}")
print(f"Input window matrix shape (X): {X.shape} -> (samples, time_steps)")
print(f"Output window matrix shape (y): {y.shape} -> (samples, time_steps)")


print("\n--- Sample Window 1 ---")
print(f"Input (last 5 values of 24h): {X[0][-5:]}")
print(f"Target (next 1h): {y[0]}")
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import GRU, Dense, Input


split_idx = int(len(X) * 0.8)

X_train, X_val = X[:split_idx], X[split_idx:]
y_train, y_val = y[:split_idx], y[split_idx:]


X_train = np.expand_dims(X_train, axis=-1)
X_val = np.expand_dims(X_val, axis=-1)

print(f"Train input shape: {X_train.shape}")
print(f"Validation input shape: {X_val.shape}")


model = Sequential([
    Input(shape=(INPUT_WIDTH, 1)),
    GRU(64, activation='tanh', return_sequences=False),
    Dense(32, activation='relu'),
    Dense(LABEL_WIDTH)
])

model.compile(optimizer='adam', loss='mse', metrics=['mae'])
model.summary()


history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=15,
    batch_size=32,
    verbose=1
)


plt.figure(figsize=(10, 5))
plt.plot(history.history['loss'], label='Train Loss (MSE)')
plt.plot(history.history['val_loss'], label='Validation Loss (MSE)')
plt.title('GRU Training and Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss (MSE)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error


predictions = model.predict(X_val)


y_true = y_val.flatten()
y_pred = predictions.flatten()


mae = mean_absolute_error(y_true, y_pred)
mse = mean_squared_error(y_true, y_pred)
rmse = np.sqrt(mse)

print("=== Model Evaluation on Validation Set ===")
print(f"Mean Absolute Error (MAE): {mae:.2f} vehicles/hour")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f} vehicles/hour")


plt.figure(figsize=(15, 6))
plt.plot(y_true[:100], label='Actual Traffic Volume', color='blue', marker='o')
plt.plot(y_pred[:100], label='Predicted Traffic Volume', color='orange', linestyle='--', marker='x')
plt.title('Actual vs Predicted Hourly Traffic Volume (First 100 Validation Steps)')
plt.xlabel('Time Step (Hours)')
plt.ylabel('Traffic Volume (Vehicles/Hour)')
plt.legend()
plt.grid(True)
plt.show()
```

```python

CONGESTION_THRESHOLD = 2200


plot_steps = 72
test_actual = y_true[:plot_steps]
test_predicted = y_pred[:plot_steps]
timestamp_index = subset_df.index[split_idx + INPUT_WIDTH : split_idx + INPUT_WIDTH + plot_steps]


flagged_congestion_times = [time for time, pred in zip(timestamp_index, test_predicted) if pred > CONGESTION_THRESHOLD]

plt.figure(figsize=(15, 7))
plt.plot(timestamp_index, test_actual, label='Actual Traffic Volume', color='blue', marker='o', alpha=0.7)
plt.plot(timestamp_index, test_predicted, label='Predicted Traffic Volume (Model)', color='darkorange', linestyle='--', marker='x', linewidth=2)


plt.axhline(y=CONGESTION_THRESHOLD, color='red', linestyle=':', label='Congestion Threshold (2200 veh/hr)', linewidth=2)

plt.fill_between(timestamp_index, CONGESTION_THRESHOLD, test_predicted,
                 where=(test_predicted > CONGESTION_THRESHOLD),
                 color='red', alpha=0.15, label='Pre-emptive Congestion Mitigation Window')

plt.title('Traffic Planning & Congestion Forecasting Dashboard (72-Hour Demonstration)', fontsize=14)
plt.xlabel('Date & Time', fontsize=12)
plt.ylabel('Traffic Volume (Vehicles/Hour)', fontsize=12)
plt.legend(loc='upper right', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.5)
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

print("=== SMART CITY TRAFFIC PLANNING DECISION SUPPORT ===")
print(f"Active Alert: Identified {len(flagged_congestion_times)} hours expecting congestion in the next 3 days.")
print("\nAction Plan:")
for time in flagged_congestion_times[:5]:
    print(f" -> [Pre-emptive Routing] Dispatch detour signals & optimize traffic lights at: {time.strftime('%Y-%m-%d %H:%M:%S')}")
```
